```python
import pandas as pd
import statsmodels.api as sm
import numpy as np
import pyarrow as pa
from build_features import build_features

train = pd.read_csv("../data/raw/Train_transactions.csv", dtype={"zip_code": str})
test = pd.read_csv("../data/raw/Test_transactions.csv", dtype={"zip_code": str})
repurchase = pd.read_csv("../data/raw/Train_customers_repurchase.csv")

X_train = build_features(train).join(repurchase.set_index("id_client"))
F_test = build_features(test)

assert len(X_train) == 160_424
assert F_test.index.is_unique
assert "repurchase" not in F_test.columns

X_train.to_parquet("../data/processed/train_features.parquet")
F_test.to_parquet("../data/processed/test_features.parquet")
```
