On fait une RFM qui permet de créer des features très intéressants pour chaque client : la date de dernier achat, la monetary value de chaque client et son panier moyen entre autre. On crée également d'autres features, sur les clients avec un comportement étrange (panier négatif etc..), sur la temporalité du client (en utilisant une variable first buy pour créer "l'âge client")

```python
ref_date = train["transaction_date"].max().normalize() + pd.Timedelta(days=1)
pos = train[train["basket_value"] > 0]

rfm = pos.groupby("id_client").agg(
    last_purchase=("transaction_date", "max"),
    first_purchase=("transaction_date", "min"),
    frequency=("transaction_date", "size"),
    monetary=("basket_value", "sum"),
    avg_basket=("basket_value", "mean"),
)
rfm["recency"] = (ref_date - rfm["last_purchase"]).dt.days
rfm["tenure"] = (ref_date - rfm["first_purchase"]).dt.days
rfm = rfm.drop(columns=["last_purchase", "first_purchase"])
```

On récupère également quelques features sur les clients ayant fait des retours. Même si ils sont peu, cela aura sûrement une influence sur la probabilité de retour

```python
train["is_return"] = train["basket_value"] < 0
ret = train[train["is_return"]]

ret_feat = ret.groupby("id_client").agg(
    nb_returns=("basket_value", "size"),
    return_amount=("basket_value", lambda s: -s.sum()),
    avg_return=("basket_value", lambda s: -s.mean()),
)
```

```python
features = rfm.join(ret_feat, how="left")
cols = ["nb_returns", "return_amount", "avg_return"]
features[cols] = features[cols].fillna(0)

n_tickets = train.groupby("id_client").size().rename("n_tickets")
features = features.join(n_tickets)

features["has_return"] = (features["nb_returns"] > 0).astype(int)
features["return_rate"] = features["nb_returns"] / features["n_tickets"]
features["return_share_amount"] = features["return_amount"] / features["monetary"]

features = features.join(repurchase.set_index("id_client"))
```
